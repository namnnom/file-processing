```python
string_cols = result.select_dtypes(include=["object"]).columns

# Функция делает первую букву заглавной, сохраняя регистр остального текста
def capitalize_first_only(val):
    if isinstance(val, str) and len(val) > 0:
        return val[0].upper() + val[1:]
    return val

# Применяем функцию к каждому элементу текстовых столбцов
result[string_cols] = result[string_cols].map(capitalize_first_only)
```

```python
def capitalize_after_semicolon(val):
  if isinstance(val, str):
    # Ищет точку с запятой, за которой следует любая буква, и делает её заглавной
    return re.sub(
        r'(;)\s*([a-zа-яё])',
        lambda m: m.group(1) + ' ' + m.group(2).upper(),
        val,
        flags=re.IGNORECASE,
    )
  return val


# Применяем ко всем столбцам в DataFrame
result = result.apply(lambda col: col.map(capitalize_after_semicolon))
```

```python
# 1. Выделяем все колонки для склеивания (кроме самого ключа группировки)
columns_to_agg = [col for col in result.columns if col != 'Код ЕПБС']

# 2. Создаем lambda-функцию с очисткой от NaN, пустых строк "" и пробелов
clean_join_lambda = lambda x: '; '.join(set(
    val for val in x.dropna().astype(str) if val.strip() != ''
))

# 3. Запускаем группировку, применяя эту функцию ко всем нужным колонкам
df_result = result.groupby('Код ЕПБС', as_index=False).agg({
    col: clean_join_lambda for col in columns_to_agg
})
#склеили НПА
```

```python
group_cols = [
    r'Услуга\работа (код услуги\работы)',
    'Код вида деятельности',
    'Признак отнесения к услуге или работе',
    'Платность',
    'Наименование содержания',
    'Наименование условия',
]
columns_to_agg = [col for col in df_result.columns if col not in group_cols]

# 1. Жесткая нормализация ключей группировки
# Список значений, которые считаем пустотой:
empty_markers = {'', 'nan', 'none', 'null', 'na', '<na>', 'nat'}

for col in group_cols:
  # Переводим всё в строку, убираем обычные и неразрывные пробелы (\xa0)
  s = (
      df_result[col]
      .astype(str)
      .str.replace('\xa0', ' ', regex=False)
      .str.strip()
  )
  # Заменяем все варианты пустот на единую строку-маркер (или np.nan)
  df_result[col] = np.where(s.str.lower().isin(empty_markers), np.nan, s)


# 2. Функция склейки значений
def clean_join(series):
  cleaned = (
      val.replace('\xa0', ' ').strip()
      for val in series.dropna().astype(str)
      if val.strip().lower() not in empty_markers
  )
  return '; '.join(dict.fromkeys(cleaned))


# 3. Группировка
df_result = df_result.groupby(group_cols, as_index=False, dropna=False)[
    columns_to_agg
].agg(clean_join)
```

```python
group_cols = [
    r'Услуга\работа (код услуги\работы)',
    'Код вида деятельности',
    'Признак отнесения к услуге или работе',
    'Платность',
]
target_col = 'Признак отнесения к услуге или работе'
base_cols = [c for c in group_cols if c != target_col]
columns_to_agg = [col for col in df_result.columns if col not in group_cols]

# 1. Жесткая нормализация ключей группировки
empty_markers = {'', 'nan', 'none', 'null', 'na', '<na>', 'nat'}

for col in group_cols:
  s = (
      df_result[col]
      .astype(str)
      .str.replace('\xa0', ' ', regex=False)
      .str.strip()
  )
  df_result[col] = np.where(s.str.lower().isin(empty_markers), np.nan, s)

# Подтягиваем непустой 'Признак...' из соседних строк, если совпадают 3 остальных поля
df_result[target_col] = df_result.groupby(base_cols, dropna=False)[
    target_col
].transform(lambda s: s.ffill().bfill())

# Список столбцов для склейки через |
pipe_cols = {'Содержание', 'Условия'}


# 2. Универсальная функция склейки значений
def make_clean_join(sep='; '):
  def clean_join(series):
    cleaned = (
        val.replace('\xa0', ' ').strip()
        for val in series.dropna().astype(str)
        if val.strip().lower() not in empty_markers
    )
    return sep.join(dict.fromkeys(cleaned))

  return clean_join


# Формируем словарь агрегации
agg_dict = {
    col: make_clean_join(' | ' if col in pipe_cols else '; ')
    for col in columns_to_agg
}

# 3. Группировка с применением словаря
df_result = df_result.groupby(group_cols, as_index=False, dropna=False).agg(
    agg_dict
)
```

```python
col_name = 'Действует с'

# 1. Разбиваем строки по '$', убираем пробелы и переводим в формат даты
dates_df = (
    df_result[col_name]
    .astype(str)
    .str.split('; ', expand=True)
    .apply(lambda s: pd.to_datetime(s.str.strip(), dayfirst=True, errors='coerce'))
)

# 2. Находим минимальную дату по каждой строке (axis=1)
df_result['Действует с'] = dates_df.min(axis=1).dt.strftime('%d.%m.%Y')
# df_result[col_name] = dates_df.min(axis=1).dt.strftime('%d.%m.%Y')
```
