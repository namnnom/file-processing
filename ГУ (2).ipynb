```python
df_1=df_1[['Код вида деятельности', 'Реестровый номер',
       'Наименование базовой услуги или работы', 'Содержание1 Наименование',
       'Содержание1 Значение', 'Содержание2 Наименование',
       'Содержание2 Значение', 'Содержание3 Наименование',
       'Содержание3 Значение', 'Условие1 Наименование', 'Условие1 Значение',
       'Условие2 Наименование', 'Условие2 Значение',
       'Признак отнесения к услуге или работе',
       'Код ОКВЭД', 'Код ОКПД',
       'Категория потребителей услуги (работы)', 'Показатели объема',
       'Показатели качества']]
```

```python
df_1 = df_1.rename(columns={'Реестровый номер': 'Номер, установленный субъектом'})
```

```python
df_3=df_3[['Уникальный номер услуги\работы', 'Наименование НПА']]
```

```python
df_3 = df_3.rename(columns={'Уникальный номер услуги\работы': 'Код ЕПБС'})
```

```python
df_2
```

```python
result = df_2.merge(df_3, on='Код ЕПБС', how='left').merge(df_1, on='Номер, установленный субъектом', how='left')
```

```python
result
```

```python
result['Наименование НПА']=result['Наименование НПА'].str.strip("'\"")
```

```python
result['Наименование НПА'] = result['Наименование НПА'].fillna('')
```

```python
df_result=result.groupby('Код ЕПБС', as_index=False).agg(lambda x: '; '.join(set(x.dropna().astype(str))))
```

```python
df_result['Номер, установленный субъектом']=df_result['Номер, установленный субъектом'].astype(str).str.zfill(23)
```

```python
df_result['Код вида деятельности']=df_result['Код вида деятельности'].astype(str).str.split('.', expand=True)[0].str.zfill(2)
```

```python
df_result['Код ОКВЭД']=df_result['Код ОКВЭД'].str.replace('\.', '', regex=True).str.replace('\n', ' ', regex=True)
```

```python
df_result['Код ОКПД']=df_result['Код ОКПД'].str.replace('\.', '', regex=True).str.replace('\n', ' ', regex=True)
```
